```python
from matplotlib.ticker import FuncFormatter

sns.set_theme(style = 'whitegrid')

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(x='target', y='credit_amount', data = df, hue = 'target', palette='RdYlGn_r', legend = False,  ax=axes[0])
axes[0].set_title('Credit Amount by Default Status', fontsize=14)
axes[0].set_xlabel('Default( 0 = No, 1 = Yes)', fontsize=11)
axes[0].set_title('Credit Amount (DM)', fontsize=11 )
axes[0].set_ylim(0, 15000)



sns.boxplot(x='target', y='age', data=df, hue = 'target', palette='RdYlGn_r', legend = False, ax=axes[1])

axes[1].set_title('Age by Default Status', fontsize=14)
axes[1].set_xlabel('Default( 0 = No, 1 = Yes)', fontsize=11)
axes[1].set_title('Age (Years)', fontsize=11 )

plt.tight_layout
plt.show()
```
