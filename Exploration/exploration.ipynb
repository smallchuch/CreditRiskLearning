```python
from matplotlib.ticker import FuncFormatter

sns.set_theme(style = 'whitegrid')

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(x='target', y='credit_amount', data = df, hue = 'target', palette='RdYlGn_r', legend = False,  ax=axes[0])
axes[0].set_title('Credit Amount by Default Status', fontsize=14)
axes[0].set_xlabel('Default( 0 = No, 1 = Yes)', fontsize=11)
axes[0].set_title('Credit Amount (DM)', fontsize=11 )
axes[0].set_ylim(0, 15000)



sns.boxplot(x='target', y='age', data=df, hue = 'target', palette='RdYlGn_r', legend = False, ax=axes[1])

axes[1].set_title('Age by Default Status', fontsize=14)
axes[1].set_xlabel('Default( 0 = No, 1 = Yes)', fontsize=11)
axes[1].set_title('Age (Years)', fontsize=11 )

plt.tight_layout
plt.show()
```

Checking credit default by employment type

```python
employment_map = {
    'A71': 'unemployed',
    'A72': '< 1 Year',
    'A73': '1 - 4 Years',
    'A74': '4 - 7 Years',
    'A75': ' >= 7 Years',
}

employment_default = df.groupby(df['employment'].map(employment_map))['target'].mean().sort_values(ascending=False)

fix, ax = plt.subplots(figsize = (10,5))

sns.barplot(x=employment_default.index,
            y=employment_default.values,
            palette = 'RdYlGn_r',
            hue=employment_default.index,
            legend=False,
            ax=ax
            )

ax.set_title("Default Rate by Employment Length", fontsize=14, fontweight='bold')
ax.set_xlabel("Employment Length", fontsize=12, fontweight='bold')
ax.set_ylabel("Default Rate", fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```
